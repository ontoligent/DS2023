# Seaborn Datasets

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
datasets = sns.get_dataset_names()
for i, ds_name in enumerate(datasets):
    print(i, ds_name)
```

```python
n = 13
print(datasets[n])
df = sns.load_dataset(datasets[n])
print(df.shape)
df.head()
```

```python
df.set_index(['Year', 'Country']).Spending_USD.unstack().plot()
plt.show()
```

```python
df.plot.scatter(x='Spending_USD', y='Life_Expectancy', c='Country')
plt.show()
```

```python
df2 = df.groupby(['Country']).agg({'Life_Expectancy':'mean', 'Spending_USD':'sum'})
df2.style.background_gradient(axis=0)
```

```python
df2.Life_Expectancy.plot()
plt.show()
```

```python
df2.Spending_USD.plot()
plt.show()
```
