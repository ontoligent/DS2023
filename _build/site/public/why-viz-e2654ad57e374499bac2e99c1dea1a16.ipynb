## Vision Trumps All

> "Vision trumps all other senses.  
> We learn and remember best through pictures, not through written or spoken words."  
> — John Medina, *Brain Rules*

## Exercise Overview

We will test this idea by comparing:
- Memory for a **data table**
- Memory for a **chart**

## Table Condition

### Memorize this table (30 seconds)

| Country   | Population (millions) | GDP (trillions $) |
|-----------|------------------------|-------------------|
| USA       | 331                    | 23.0              |
| China     | 1441                   | 17.7              |
| Japan     | 126                    | 4.9               |
| Germany   | 83                     | 4.2               |
| India     | 1380                   | 3.1               |
| Brazil    | 213                    | 1.6               |

**Speaker Notes:**  

- Give students **30 seconds** to study the table.  
- Then hide the slide.  
- After ~1 minute, ask recall questions:
  - Which country had the largest GDP?  
  - Which two countries had nearly equal GDPs?  
  - Roughly how many millions of people live in Japan?  

## Recall Questions (Table)

- Which country had the largest GDP?  
- Which two countries had nearly equal GDPs?  
- Roughly how many millions of people live in Japan?  

## Chart Condition

### Memorize this chart (30 seconds)

```r
library(ggplot2)

df <- data.frame(
  Country = c("USA","China","Japan","Germany","India","Brazil"),
  Population = c(331,1441,126,83,1380,213),
  GDP = c(23.0,17.7,4.9,4.2,3.1,1.6)
)

ggplot(df, aes(x = Population, y = GDP, label = Country)) +
  geom_point(size = 4, color = "steelblue") +
  geom_text(vjust = -1) +
  labs(x = "Population (millions)", y = "GDP (trillions $)",
       title = "Population vs. GDP by Country") +
  theme_minimal()
```

````python
import plotly.express as px
import pandas as pd
````

````python
df = pd.DataFrame(
    dict(
        Country = ["USA","China","Japan","Germany","India","Brazil"],
        Population = [331,1441,126,83,1380,213],
        GDP = [23.0,17.7,4.9,4.2,3.1,1.6]
    )
)
````

````python
df
````

````python
fig = px.scatter(df, 'Population', 'GDP', text='Country', height=600, width=700)
fig.update_traces(textposition="top center", marker=dict(size=14), textfont=dict(size=14))
fig.update_layout(margin=dict(l=20,r=20,t=60,b=20))
fig
````
