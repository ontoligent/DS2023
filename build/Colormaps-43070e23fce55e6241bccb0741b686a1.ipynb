## Overview

In Matplotlib, a colormap is used to **map** scalar **data** values **to color** values.

Colors are used to **replace numbers** in a visualization to display quantitive meaning.

Data are usually **normalized to a scale** between $0$ and $1$ and then **mapped** onto colors represented using the **RGBA** encoding scheme.


## Use cases

Colormaps are used **everywhere** in visualization.

Nearly **every plot type** in Matplotlib has a paramter to define color.

Colormaps are especially useful for visualizing 2D or 3D data arrays in plots like **heatmaps**, surface plots, **scatter plots**, and contour maps.

## Representing color as RGBA

RGBA is a color encoding scheme that defines a color in terms of four channels: **R**ed, **G**reen, **B**lue, and **A**lpha.

Each color channel may have a value between $0$ and $255$, yielding $256^3$ or $16{,}777{,}216$ colors.

In this way, every color is defined as a mixture of red, green, and blue.

The alpha channel determines the opacity of a color, ranging from $0.0$ to $1.0$.

## Palettes and their uses

Matplotlib comes with a number of colormaps to choose from.

These are grouped into **five primary classes** based on their **mathematical structures** and **how human eyes perceive** them.

### Perceptually Uniform

- General scientific data, accurate perception.
- Lightness increases monotonically; converts flawlessly to grayscale.

<img src="https://matplotlib.org/stable/_images/sphx_glr_colormaps_001.png"/>

## Sequential

- Ordered data moving from low to high values.
- Smooth variation between two color tones (e.g., low saturation to high saturation).

<img src="https://matplotlib.org/stable/_images/sphx_glr_colormaps_002.png"/>
<img src="https://matplotlib.org/stable/_images/sphx_glr_colormaps_003.png"/>

## Diverging

- Data with a critical midpoint or zero value.
- Two divergent hues that meet at an unsaturated, neutral middle color.

<img src="https://matplotlib.org/stable/_images/sphx_glr_colormaps_004.png"/>

## Cyclic

- Wrapping values or periodic data.
- Start and end colors match smoothly; ideal for phase angles, wind directions, or hourly time.

<img src="https://matplotlib.org/stable/_images/sphx_glr_colormaps_005.png"/>

## Qualitative

- Unordered or categorical classes, as with nominal grouping data.
- Distinct, miscellaneous hues with non-monotonic lightness to distinguish discrete sets.

<img src="https://matplotlib.org/stable/_images/sphx_glr_colormaps_006.png"/>

## Putting colormaps to work

Let's put the colormap to work.

In the bad old days, Pandas' scatter plot function did not allow you to pass a variable name to the `c` argument. 

To get that effect, you had to do it by hand.

Tedious as it was, it illustrated how colormaps can be used.

This is useful knowledge in other contexts.

**Let's look at an example.**

Let's create a scatter plot of two variables from the `iris` dataset and color them by `species`.

````python
import seaborn as sns
import pandas as pd

iris = sns.load_dataset('iris')
iris.head()
````

To this, we first generate a number to go with each member of the species domain.

If the variable is cast as a `Categorical` data type, this is easy: we just grab the `codes` associated with the elements (levels) of the category.

````python
iris['species'] = pd.Categorical(iris['species'])
iris.species.cat.codes
````

Now all we do is pass the series of codes to the `c` argument.

We also specify a colormap with `cmap`.

Here we choose on from the **qualitative** group.

````python
iris.plot.scatter(
    x='sepal_length', 
    y='petal_width', 
    c=iris.species.cat.codes, 
    cmap='Set1')
plt.show()
````

The result is not ideal because the colorback represents the number as a continuous variable, and does not show the relationship between color and category.

To get the proper result, we turn off the colorbar and add a legend.

````python
ax = iris.plot.scatter(
    x='sepal_length', 
    y='petal_width', 
    c=iris.species.cat.codes, 
    cmap='Set1',
    colorbar=False # Turn off the colorbar
)

# Add a legend
ax.legend(
    handles=ax.collections[0].legend_elements()[0],
    labels=list(iris['species'].cat.categories),
    title="Species"
)

plt.show()
````

Again, this is tedious, but you can see how colormaps are used.

They take a numeric represenation of variable and return a color that can be used for visualization.

## Color names

Finally, here are some of the color names available to use in Matplotlib whenever you are asked to define a color for some visual feature.

For example, you may want to define the color of a line as magenta.

Matplotlib comes with a few libraries of colors:

- BASE_COLORS
- CSS4_COLORS
- TABLEAU_COLORS
- XKCD_COLORS

The names and their color values are available as dictionaries from the `mpl.colors` object.

For example:

````python
mpl.colors.BASE_COLORS
````

````python
mpl.colors.TABLEAU_COLORS
````

Let's create a function to display a color given its name.

````python
import matplotlib.patches as patches

def plot_swatch(color_name):
    fig, ax = plt.subplots(figsize=(2, 2))
    swatch = patches.Rectangle((0, 0), 1, 1, facecolor=color_name)
    ax.add_patch(swatch)
    ax.set_xlim(0, 1)
    ax.set_ylim(0, 1)
    ax.axis('off')
    ax.set_title(color_name)
    plt.show()
````

````python
plot_swatch('tab:orange')
````

Here is a function to show all the colors in a collection.

````python
def print_colors(mpl_library, w=8):
    
    N = len(mpl_library)
    h = N // w + int(bool(N % w))

    fig, ax = plt.subplots(h, w, figsize=(w, h*1.1), subplot_kw={'xticks': [], 'yticks': []})
    
    for i, color_name in enumerate(mpl_library):
        
        c = i % w
        r = i // w
        
        swatch = patches.Rectangle((0, 0), 1, 1, facecolor=color_name)
        ax[r,c].add_patch(swatch)
        ax[r,c].set_title(color_name.replace(":","\n"), fontsize=8)
        sns.despine(left=True, bottom=True)

    plt.tight_layout()
    plt.show()
        
````

````python
print_colors(mpl.colors.BASE_COLORS, w=4)
````

````python
print_colors(mpl.colors.TABLEAU_COLORS, w=5)
````

````python
print_colors(mpl.colors.CSS4_COLORS)
````

````python
color_dictionary = pd.Series(mpl.colors.BASE_COLORS | mpl.colors.CSS4_COLORS | mpl.colors.TABLEAU_COLORS | mpl.colors.XKCD_COLORS)
len(color_dictionary)
````

````python
print_colors(color_dictionary.sample(64).to_dict())
````
