## Overview

Scatter plots are designed to show the relationship between two numeric features.

Specifically, they can show the correlations

## A simple example

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

Let's generate some random numeric data

```python
df = pd.DataFrame(
    {
        'f1': np.random.randn(1000),
        'f2': np.random.randn(1000)
    }
)
df.head()
```

```python
df.plot.scatter(x='f1', y='f2')
plt.grid()
plt.show()
```

## A more complex example

```python
penguins = sns.load_dataset('penguins').dropna()
penguins
```

```python
# penguins.info()
```

```python
penguins.plot.scatter(x='bill_length_mm', y='bill_depth_mm')
plt.show()
```

```python
penguins.plot.scatter(x='bill_length_mm', y='bill_depth_mm', c='species')
plt.show()
```
