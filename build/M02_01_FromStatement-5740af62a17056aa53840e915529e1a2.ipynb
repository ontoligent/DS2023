## A more complex example

Let's go through this process again, with a slightly more complex example.

Consider the following descriptive statement, a sentence you might find in a report on global health.

> "In 1970, Germany spent $USD 252.311 per capita and had a life expectany of 70.6 years."

Now, this sentence can be decomposed into a set of atomic facts.

We might write down these facts like this:

```
Observation1 country Germany
Observation1 year 1970
Observation1 spending 252.311
Observation1 life_expectancy 70.6
```

Note that in writing these doown we have surfaced something interesting -- the implicit subject of the sentence is the observation. 

This is a little tricky, because it looks like the the subject is the country. But the year 1970 is not a property of Germany!

Now, imagine we have a bunch of these sets of facts, one set for each country in the world and year between 1970 and 2020.

A segment of these data might look like the following multi-line Python string:

````python
data = """
Observation1 country Germany
Observation1 year 1970
Observation1 spending 252.311
Observation1 life_expectancy 70.6
Observation2 country France 	
Observation2 year 1970 	
Observation2 spending 192.143 	
Observation2 life_expectancy 72.2
Observation3 country GreatBritain
Observation3 year 1970
Observation3 spending 123.993
Observation3 life_expectancy 71.9
Observation4 country Japan
Observation4 year 1970
Observation4 spending 150.437
Observation4 life_expectancy 72.0
Observation5 country USA
Observation5 year 1970
Observation5 spending 326.961
Observation5 life_expectancy 70.9 
"""
````

## Putting the facts into a table

Let's use Pandas to convert these facts into a table

We'll use a list comprehension to parse the string into a list of lists and pass that to a data frame constructor.

````python
lol = [row.split() for row in data[1:-1].split("\n")]
lol
````

````python
import pandas as pd

df = pd.DataFrame(lol, columns=['Subject','Predicate','Object'])
df
````

## Reshaping the table

Now, the shape of this table is not how we normally want our data.

It shows how the data set is composed of facts, but it's not useful in this form for data _analysis_.

For one thing, the Object column contains data of different types, which prevents storing them and processing them properly.

One thing we can do is pivot the table into this form:

````python
df1 = df.pivot(index='Subject', columns='Predicate', values='Object')
df1
````

We see that that the pivot operation sorted the columns alphatically. 

A more intuitive way of represented the data is as follows:

````python
df1[['country','year','spending','life_expectancy']]
````

In this version of the table, the columns show the common pattern of the descriptive sentence: "Country $A$ in year $B$ spent $C$ dollars per capita and had a life expectancy of $D$." The sentence encodes a hypothesized cause and effect relationship: what factors influence life expectancy?

> [!important] This puts our data into "tidy" form -- each row is a unique observation, and each column is unique feature, or variable.

## Summary

We have seen that a table can be constructed from a collection of **atomic facts**.

Each atomic fact has a **three-part structure**: Subject, Predicate, Object.

The subject of an atomic fact is often **observation** itself.

These parts can be **mapped onto a table** as rows, columns, and cells.

Tables are a **primary form of visualizing** linguistic information.

> [!important] The lesson here is that to understand a table, you first need to understand that implied structure of the atomic facts it represents. To identify the proper subject of these facts is to understand what the data are about.
