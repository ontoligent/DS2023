## Chart Condition

### Memorize this chart (30 seconds)

```python
# import plotly.express as px
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_theme(style='darkgrid')

df = pd.DataFrame(
    dict(
        Country = ["USA","China","Japan","Germany","India","Brazil"],
        Population = [331,1441,126,83,1380,213],
        GDP = [23.0,17.7,4.9,4.2,3.1,1.6]
    )
).set_index('Country')

ax = df.plot.scatter(x='Population', y='GDP', figsize=(10,10))
for idx, row in df.iterrows():
    y_pad = 0 if idx != 'Germany' else -.1
    ax.text(row['Population']+20, row['GDP']+y_pad, idx)
sns.despine()
plt.show()
```
