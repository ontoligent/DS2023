## Data Types

Most programming languages and databases support a variety of **data types**.

Here are some data types supported by Python:

1. Character strings `str`, e.g. `'this is a string'`
3. Integers `int`, e.g. `1`, `205`
4. Real numbers `float`, e.g. `.06`, `101.0`
5. Complex numbers `complex`, e.g. `5 - 2j` where `j` is imaginary
6. Boolean values `bool`, e.g. `True`, `False`

## Why have data types?

A data type determines the kind of **operations** that can be performed on the data.

For example, integers and other numeric data types may be added, subtracted, divided, etc.

Strings can be concatenated, spliced, and combined in various ways to form new strings.

Boolean data can be combined and compared to compute truth values.

## There are other types of data

There are other types, and variants of these, but these are the main types.

Data types are related to but should not be confused with **data structures**.

In Python, the common data structures are lists, tuples, sets, and dictionaries.

The Pandas library introduces new data structures to Python: series and dataframes.

We'll discuss these later.

## The uses of data

Just as data types determine the kinds of operations that can be performed on data, data types can be classified by the purpose they have within a data set.

There are two main uses of data in a data set, in particular in tidy data:

- **To identify** an observation.
- **To measure** an attribute of an observation.

Example of identifiers are proper names. 

Example of measures are height, weight, velocity, etc. -- basically anything that a scientific instrument is designed to capture and record in a data set.

> [!note] Measures don't have to be numbers -- they can be categories that classify an observation into a group.
>
> Also, sometimes combinations of categories, called "keys," can be used to perform the work of identification.
>
> All of this will become clearer with examples, which we will delve into later.
