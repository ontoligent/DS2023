## Filtering rows

One of the most important operations to perform on a dataframe is filtering.

Filtering grabs a subset of rows based on some specified conditions.

> [!important] Filtering is a **row-wise operation**: We use it to extract a subset of **rows** that meet certain conditions.

For example, we may want to see only rows for a certain year and month.

### The `query` method 

We can use `.query()` to do this. It works by passing a condition expressed as a string. 

For example, to see all flights for February 2013, we would do this:

````python
flights.query("year == 2013 & month == 2")
````

Here is a query for flights delayed more than 120 minutes:

````python
flights.query("dep_delay > 120")
````

### The indexer method

You also filter by supplying the two expressions directly to the dataframe's **indexer**, which you can access by using two square brackets at the end of the variable name, e.g. `flights[]`.

Note that the two boolean expressions are enclosed in parentheses.

````python
flights[(flights['year'] == 2013) & (flights['month'] == 2)]
````

### The `loc` accessor method

You can also query by using Pandas' `loc` accessor method to do queries.

````python
flights.loc[flights['dep_time'] > 120]
````

## Sorting rows with `sort_values`

The method `.sort_values()` changes the order of rows based on the value of one or more columns. 

It takes a list of column names, or a single column name, to sort by.

The `ascending` argument determines the resulting sort order, whether from least to greatest (`asending = True`) or not. The default value is `True`, so if we want reverse sort order we set it to `False`.

````python
flights.sort_values('dep_delay', ascending=False)
````

Note that the number of rows hasn't changed. Sorting rearranges the data, it doesn't filter it.

`.drop_duplicates()` finds unique rows in a dataset.

````python
flights1 = flights.drop_duplicates()
````

## Creating new columns

You can create new columns by naming them in the indexer and then assigning a value to them.

Let's compute the `gain` (how much time a delayed flight made up in the air) and the `speed` in miles per hour:

````python
flights['gain'] = flights['dep_delay'] - flights['arr_delay']
flights['speed'] = flights["distance"] / flights["air_time"] * 60
flights
````

> [!note] These new columns are what we have been called "analytical" columns. They are direct transforamtions of the data and don't add any new empirical information to the dataset.

## Selecting columns

Another important operation to perform on a dataframe is to select a subset of columns.

Sometimes you do this because there are too many columns to see on one page. Other times, you want to zoom in on a few columns for exploratory purposes.

> [!important] Selecting is a **column-wise** operation. We use it to select a subset of columns from the dataframe.

To select columns, we just pass a list of column names to the dataframe.

````python
flights[["year", "month", "day"]]
````

Note that we can pass a list of columns in a different order than how they appear in the dataframe.

````python
flights[["year", "day", "month"]]
````

We can also select a *range* of columns by position with `loc` method.

The `:` in the expression below means all rows &mdash; we'll look at this function more below.

````python
flights.loc[:, "year":"day"]
````

We can use our column lists from above to select, too.

````python
flights[group_cols]
````

````python
flights[measure_cols]
````

## Combining filtering and selecting with `loc`

With the `loc` method, you can combine filtering and selecting.

- In the **first argument**, you can put a **filtering** expression. This is usually a boolean expression applied to one or more columns in the dataframe. 

- In the **second argument**, you can put a **selecting** expression. The is usually a list of columns names.

Think of it this way:

```python
df.loc[rowwise_expression, columnwise_expresson]
```

For example:

````python
flights.loc[flights.month == 1, ['dep_time','dep_delay']]
````

## Computing stats

The whole point of a dataframe is to be able to compute statistical and other mathematical properties from the data.

Here are a couple of quick things you can do.

### Get the number of values in a column or combination of columns with `value_counts`

You can use `value_counts()` to quickly count the number of rows associated with one or more columns:

````python
flights.value_counts('carrier')
````

````python
flights.value_counts(['origin','dest'])
````

Note that `value_counts` returns a series , not a dataframe.

If want a dataframe, we can use the `.to_frame()` method.

````python
flights.value_counts(['origin','dest']).to_frame('n')
````

Note also the `.value_counts()` returns a series with an index for the counted columns.

When we have at least two grouping columns in the index, unstack the results, since the count is a measurement.

Here we see a wide table showing the top 20 most frequent origin and destination combinations.

````python
flights.value_counts(['origin','dest']).head(20).unstack(fill_value=0)
````

### Computing aggregate values

You can compute aggregate values in many ways in Pandas. Here are two of the most common.

One is to just apply the aggregate function to one or more columns.

Here we get the mean value of each of our measurement columns.

````python
flights[measure_cols].mean()
````

We can also pass the aggregation funcfion name to the `.agg()` method.

````python
flights[measure_cols].agg('mean')
````

The `.agg()` method has the advantage of being able to take more than one function.

Here we get the mean, median, and standard deviation. 

````python
flights[measure_cols].agg(['mean','median','std'])
````

We can transpose the table with the `.T` operator to turn the columns into observations.

````python
flights[measure_cols].agg(['mean','median','std']).T
````

### Using `.groupby()`

We can also use the `.groupby()` method to compute joint aggregate values on combinations of columns.

Here we get the mean values for all of our measure columns for each carrier to each airport.

````python
flights.groupby(['carrier','dest'])[measure_cols].mean()
````

Note that `.groupby()` also returns a series with an index.

In this case, we can unstack the results on any of our aggregate measures, like so:

````python
flights.groupby(['carrier','dest'])['dep_delay'].mean().unstack(fill_value=0).style.format(precision=0)
````

## Chaining it all together

You've now seen several individual operations.

It is possible to combine these operations.

For example, if you wanted to find the fastest flights to Charlottesville's CHO airport: you need to combine filtering, creating a new column, selecting columns, and sorting. 

All in one pipeline:

````python
flights.query('dest == "CHO"')\
    .groupby(["year", "month", "day"])\
    .agg(arr_delay=("arr_delay", "mean"))\
    .reset_index()\
    .head(10)
````

Note the use of backslashes. These allow us to write a single Python expression over multiple lines.

Instead of using backslashes, we can just wrap everything in parentheses.

````python
(
    flights
    .query('dest == "CHO"')
    .groupby(["year", "month", "day"])
    .agg(arr_delay=("arr_delay", "mean"))
    .reset_index()
    .head()
)
````

Finally, here is the same set of operations using the indexer syntax.

````python
(
    flights[flights['dest'] == 'CHO']
    .groupby(["year", "month", "day"])['arr_delay']
    .mean()
    .to_frame('arr_delay')
    .reset_index()
    .head()
)
````

> [!tip] You will find that in Pandas, the principle of TMTOWTDI &mdash; There's More Than One Way To Do It &mdash; applies.
