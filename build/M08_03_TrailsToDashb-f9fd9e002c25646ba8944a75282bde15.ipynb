## Overview

This notebook describes how to **convert our insights** about actions and trails **into the design** of dashboards and other compositional data visualizations. 

To make the conversion, we pursue an **ontology-driven design** process: we take advantage of the ontological classification of actions in the previous notebook to organize and project sequences of actions onto functional visual artifacts.

This process begins with understanding the relationship betweeen **actions and trails**.

## Actions and Trails

A trail is a **sequence of actions** that leads to a **discovery**.

A trail can be composed of a series of small discoveries that lead to a bigger discovery &mdash; the length of the trail is function of the scope of your question.

Ideally, a data visualization tool would **record** these trails and **share** them for reproducibility.

- Jupyter notebooks do this already (to some extent).

- This something like a clickstream for a website, but optimized for science.

In addtion, a tool, such as a dashboard, should **reflect and support trails** of actions.

## An example trail

Consider the following sequence of actions following in the EDA notebook on wine reviews.

This is a trail because it is a series of actions that leads to a discovery. 

It also has smaller discoveries along the way.

**The Task**

The overarching task is the following:

> Find the best-value wines in a large international review dataset —
wines that score well for their price.

This is a rephrased question about the relationship between market price and expert assessment.

To break it down into subtasks, we focus on the fact that the main task requires **understanding the structure of the data** before any judgments can be made.

**$S_1$ Characterize coverage by country**

- The analyst **loads** the CSV file and **inspects** its schema and missing-value pattern.
- She then **counts** reviews by country and **visualizes the distribution** as a bar chart on a log scale.
- She **discovers** that the distribution follows a **power law**, where a small number of countries accounting for the vast majority of reviews.
- She **defines** an IQR-based threshold to **identify** the outliers, then **filters** the working dataset to yield the top seven countries: US, France, Italy, Spain, Portugal, Chile, and Argentina.

**$S_2$ Characterize taster-country co-occurrence**

- The analyst **aggregates** reviews by taster and country jointly, then **recodes** null taster names as *Anonymous*.
- She **visualizes** the result as an **alluvial** diagram and then as a **heatmap**.
- She **discovers** tasters specialize &mdash; Roger Voss concentrates on France and Portugal; the anonymous mass concentrates in the US and Italy.

**$S_3$ Explore the price-quality relationship**

- The analyst **computes** the joint frequency of price and points values and **displays** it as a **heatmap**.
- This **reveals** an unexpected **anomaly**: prices cluster at five-dollar intervals, a market structure pattern invisible to histograms and density plots.
- She then **plots** price against points as a scatterplot, **applies a log transform** to spread the crowded low-price observations, and **fits** a linear regression.
- The **result** is a weak upward trend with high variance, confirming that many wines are over- or under-priced relative to their scores.

**$S_4$: Define and identify bargain wines**

- The analyst **defines** a *bargain* using three conditions:
above the regression line, at or below median price, and at or above median score.
- She **applies** these as row labels, then **derives** a complementary metric the Bang-for-Buck Index (BBI = points ÷ price) — and uses a box plot to group wines into three BBI tiers.
- A third criterion, *cheapest in class* (price ≤ $100, points ≥ 90), rounds out the metric set.

**$S_5$ Compare countries on value**

- For each metric, the analyst **computes** the relative frequency of qualifying wines by country and **displays** the results as horizontal bar charts.
- Pie charts prove inappropriate (relative frequencies do not sum to a meaningful whole) and are corrected.
- She **discovers** that Portugal appears as the top country for best-value wines across multiple metrics; Spain dominates BBI Group 1.

**Final Discovery**

> For a wine buyer seeking value, Portugal and Spain offer the highest concentration of wines that punch above their price point.

## Translating trails

Now, if we want to translate this trail into a dashboard or infographic, we need some way to map steps onto plots, widgets, and layout.

The key insight is that **the levels we used to classify the intent of an action translate into specific kinds of visual elements**. 

Each action type has a direct dashboard counterpart determined by its level:

| Level | Verb type | Dashboard element | Examples |
| --- | --- | --- | --- |
| Meta | Goal / interpretation | Title or insight callout | *Ideate*, *Communicate* |
| Abstract | Design intent | Layout directive (sets grouping variable) | *Compare*, *Contrast* |
| Compound | Multi-step view | Panel (chart) | *Characterize Distribution*, *Find Anomalies* |
| Atomic | Filter / select | Widget | *Filter*, *Select*, *Bin* |
| Atomic | Transform | Backend process | *Normalize*, *Aggregate*, *Reduce* |

Given this mapping, we can represent a trail as movement through both steps and levels:

| Step | Level | Action | Event | Dashboard Element |
|------|-------|--------|-------|-------------------|
| 1 | Meta | Ideate | Question: *Which countries produce the best-value wines?* | Title |
| 2 | Atomic | Inspect | Schema and null pattern confirmed | — |
| 3 | Compound | Characterize Distribution | Power law; top 7 countries identified | — |
| 4 | Atomic | Filter | Working dataset scoped to top 7 countries | Country selector (dropdown) |
| 5 | Compound | Characterize Co-occurrence | Tasters specialize by region | Heatmap panel |
| 6 | Compound | Find Anomaly | Prices cluster at $5 intervals | Price distribution panel |
| 7 | Compound | Model | Log-linear trend; high variance in scores | Price vs Points panel |
| 8 | Abstract | Derive Criteria | Bargain, BBI, and CIC criteria defined | — (layout directive) |
| 9 | Atomic | Select | Bargain and BBI wines labeled | Price slider; Points slider |
| 10 | Compound | Compare | Portugal leads; Spain wins BBI Group 1 | Best Buys by Country panel |
| 11 | Meta | Communicate | Portugal = best-value country overall | Insight callout |

## Inferring layout order

The trail sequence suggests a layout order:

- The syntagmatic axis (x) gives you the reading order of the dashboard.

- Upstream steps set context for downstream steps:

- Widgets that control data (Atomic filter steps) should appear **above or before** the panels that consume that data.
- The dependency structure of the trail &mdash; which step's output feeds which later step &mdash; is the wiring diagram of the dashboard's interactivity.
