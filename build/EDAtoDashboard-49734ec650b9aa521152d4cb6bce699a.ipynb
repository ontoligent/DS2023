## Overview

This document recovers the analytical intent behind each named figure in [the EDA Demo notebook](../../Unit%2006%20Presentation/EDADemo/EDADemo.ipynb) in order to describe design logic for converting the notebook into a dashboard.



For each figure it traces the exploratory trail that produced it, names the discovery, reconstructs the task abstraction hierarchy, and derives the dashboard implications. 

The goal is to replace intuition-driven layout choices with decisions grounded in what each figure is actually for.

## Figure 1: Country Frequencies

### Trail


The analyst began by loading the full dataset (130k reviews) and scanning for missing values.
 
Discovered substantial gaps in `taster` ($~20\%$) and `price` ($~7\%$).

Next, the analyst asked about which countries contribute enough reviews to support analysis.

Produced a ranked frequency table and a box plot of those frequencies.

Discovered a sharp power-law shape — most countries appear only a handful of times.

The analyst applied an IQR-based outlier threshold ($Q3 + 1.5 × IQR$).

Identified the top 7 countries (US, France, Italy, Spain, Portugal, Chile, Argentina) as statistical outliers — not exceptions to discard, but the dominant mass of the data.

The analyst then reduced the working dataset to 116,745 reviews.

### Discovery

Country frequency follows a power law so steep that the top 7 countries are outliers by the IQR rule
yet account for the large majority of content. 

The log scale is necessary to see any structure in the
tail. 

The IQR threshold provides a principled, data-driven cutoff rather than an arbitrary one.

### Task Abstraction

**Task**: What is the geographic scope of this dataset, and which countries have enough coverage to
support reliable analysis?

**Sub-task**: Which countries have review counts high enough to be treated as well-sampled?

**Operation**: Find Extremum + Country Frequency (IQR outlier threshold as the cutoff criterion)

### Dashboard Implication

This figure becomes a global scope panel — a country selector that governs which countries appear in
all downstream panels. 

The IQR threshold should be exposed as a slider (minimum review count), letting
the user tighten or loosen the definition of "well-sampled." 

A horizontal bar chart with the threshold
line drawn dynamically is the right form. 

Selecting or deselecting countries here propagates to every
other panel.
