## Overview

Stacked bar charts can be used to visualize discrete distributions.

This example visualizes the result of a survey in which people could rate
their agreement to questions on a five-element scale.

The horizontal stacking is achieved by calling `.barh()` for each category and passing the starting point as the cumulative sum of the already drawn bars via the parameter ``left``.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
category_names = ['Strongly disagree', 'Disagree', 'Neither agree nor disagree', 'Agree', 'Strongly agree']
results = {
    'Question 1': [10, 15, 17, 32, 26],
    'Question 2': [26, 22, 29, 10, 13],
    'Question 3': [35, 37, 7, 2, 19],
    'Question 4': [32, 11, 9, 15, 33],
    'Question 5': [21, 29, 5, 5, 40],
    'Question 6': [8, 19, 5, 30, 38]
}
df = pd.DataFrame(results, index=category_names).T
```

```python
df
```

## Solution

```python
# Create a horizontal bar chart
df.plot.barh(stacked=True, legend=True, cmap='RdYlGn', figsize=(10,5))

# Reshape the legend
plt.legend(ncols=df.shape[1], bbox_to_anchor=(0, 1), loc='lower left', fontsize='small')

# Place numbers on bar segments
for y, row in enumerate(df.values):
    x_pos = 0 # Initialize for each row
    for i, x in enumerate(row):
        x_pos += x
        font_color = 'gray' if i == 2 else 'white'
        plt.text(x_pos - ((x/2) + .5), y - .07, str(x), c=font_color, fontweight='bold')

# Remove spines
sns.despine(left=True, bottom=True)

# Remove x axis ticks
plt.xticks([])

plt.show()
```
