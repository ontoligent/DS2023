## Overview

Before choosing a chart type or writing any code, a visualization designer should answer a more fundamental question: **what is the user trying to accomplish?**

**Task abstraction** is the practice of describing user goals at multiple levels of specificity — from a broad analytic purpose down to the precise data manipulations that purpose requires.

Task abstrction helps you design **better visualizations**, both static and interactive.

A chart, dashboard, or infographic built around clearly stated questions is easier to design, build, and evaluate than one built by intuition.

## Visual Analytic Activity

## The Task Hierarchy

Gotz and Zhou (2008) organize visual analytic activity in four tiers, from most abstract to most concrete.

| Tier | What it describes |
|------|-------------------|
| **Task** | A high-level analytic goal, stated in domain terms |
| **Sub-task** | A specific analytical question that contributes to the task |
| **Action** | A concrete analytic move — a data or visualization operation |
| **Event** | A single software interaction that carries out the action |

This course adds an **Operation** tier between Sub-task and Action.
An operation expresses a sub-task as an Action–Target pair, using Amar et al.'s analytic primitives as a controlled vocabulary of actions.
This gives sub-tasks a precise, implementation-neutral form before they are translated into code.

For your project proposal you will work with the Task, Sub-task, and Operation tiers.

## Task Level

A task-level question describes what a user wants to accomplish in the data.

A well-formed task-level question is:

- **Independent of any specific chart or interaction.** Do not name a chart type, a widget, or a gesture.
- **Meaningful to a domain expert.** A public-health researcher should recognize it as a real question without knowing what a scatter plot is.
- **Open-ended enough to allow more than one design.** If only one chart obviously answers it, the question is probably a sub-task in disguise.

```{warning}
**Avoid:** "Can the user select a country from a dropdown to see a bar chart of spending over time?"

This question names a widget (dropdown) and a chart type (bar chart).
It skips past the analytic goal and lands directly on a design decision.
```

```{tip}
**Aim for:** "How has healthcare spending changed across countries over the past two decades?"

This describes the analytic goal without committing to any design.
It could be answered with a line chart, a heat map, a choropleth, or a small-multiples display.
```

## Sub-Task Level

A sub-task breaks a high-level question into a specific analytical step.

Each sub-task is a concrete analytical question — one angle on the task that a single chart or computation could address.
It should be stated in plain language, without naming any chart type or widget.

Two sub-tasks for the same task must approach it from different angles.

### Writing a Good Sub-Task

A good sub-task is specific enough that a single chart or computation could answer it,
but abstract enough that it names no chart type or widget.

If it names a bar chart, a dropdown, or a filter button, it has slipped into design. Push it back up.

### Worked Example

**Task:** How does a country's healthcare spending relate to its citizens' life expectancy?

| | Sub-task question |
|---|---|
| Sub-task 1 | Which countries achieve high life expectancy despite low spending? |
| Sub-task 2 | Does the relationship plateau above a certain spending threshold? |

Both sub-tasks address the same task from different angles.
The first focuses on exceptions; the second on the shape of the overall relationship.

## Action Level

Actions, also called **operations**, are the specific data questions that serve a sub-task.

Each operation is expressed as an **Action–Target pair**, since every action is directed toward something, called a target.

- The **Action** itself is drawn from a vocubulary, such as Amar et al.'s ten analytic primitives (see table below).
- The **Target** is the variable or data feature the action is applied to.

```{tip} Connection
Action and Target occupy the positions of **verb** and **object** respectively for an atomic statement, such as we saw earlier in this course.

The implicit subject in these statements is the sub-task, as in "Sub-task $S_1$ has action $a$ on target $t$."
```

Note the two operations for the same task must use different primitives.

### Actions: Amar et al. Analytic Primitives

| Action | The question it answers |
|--------|-------------------------|
| **Retrieve Value** | What is the exact value of this item? |
| **Filter** | Which items satisfy a condition? |
| **Compute Derived Value** | What is an aggregate — a sum, average, or ratio? |
| **Find Extremum** | What is the maximum or minimum? |
| **Sort** | In what order do these items rank? |
| **Determine Range** | What are the minimum and maximum values overall? |
| **Characterize Distribution** | What is the overall shape of these values? |
| **Find Anomalies** | What is unusual or unexpected? |
| **Cluster** | Which items are most similar to one another? |
| **Correlate** | Do two variables move together? |

### Targets

The Target names the data feature the action is applied to.
Common targets:

| Target | What it refers to |
|--------|-------------------|
| **Outliers** | Values far from the norm |
| **Trends** | Patterns of change over time or across a dimension |
| **Distributions** | The spread and shape of a set of values |
| **Clusters** | Groups of similar items |
| **Extrema** | The maximum or minimum of a set |
| **Correlations** | Relationships between two or more variables |

### Worked Example

**Task:** How does healthcare spending relate to life expectancy?

| Sub-task | Operation question | Action        | Target |
|----------|--------------------|---------------|--------|
| Which countries achieve high life expectancy despite low spending? | Which countries spend less than $1,500 per capita on healthcare? | Filter | Spending |
| Does the relationship plateau above a certain threshold? | What is the average life expectancy at each spending level? | Compute Derived Value | Life Expectancy |

The two operations use different Amar primitives.
Filter selects a subset of items; Compute Derived Value produces an aggregate across groups.

## The Full Picture

Here is a complete hierarchy for one example.

| Level | Question | Form |
|-------|----------|------|
| **Task** | How does healthcare spending relate to life expectancy? | Plain domain question |
| &nbsp;&nbsp;**Sub-task 1** | Which countries achieve high life expectancy despite low spending? | Specific analytical question |
| &nbsp;&nbsp; &nbsp;&nbsp;**Operation 1** | Which countries spend less than $1,500 per capita? | Filter + Spending |
| &nbsp;&nbsp;**Sub-task 2** | Does the relationship plateau above a certain spending threshold? | Specific analytical question |
| &nbsp;&nbsp; &nbsp;&nbsp;**Operation 2** | What is the average life expectancy at each spending level? | Compute Derived Value + Life Expectancy |

Each level is more specific than the one above it.

The two sub-tasks approach the task from different angles.

The two operations use different Amar primitives.

## Summary

- A **task** is a high-level analytic goal, stated independently of any chart or interface.
- A **sub-task** is a specific analytical question — one concrete angle on the task.
- An **operation** is an Action–Target pair, where the Action is an Amar et al. analytic primitive and the Target is the data feature it is applied to.
- The two sub-tasks for the same task must approach it from different angles.
- The two operations for the same task must use different Amar primitives.

## Questions

1. A student writes this task-level question: "Can users hover over a point to see a country's exact GDP?" What is wrong with it, and how would you revise it?

2. Write a task-level question for a dashboard about college admissions. Then write two sub-tasks that approach it from different angles.

3. For each of your sub-tasks from question 2, write one operation as an Action–Target pair. Use a different Amar primitive for each.

4. What is the difference between **Find Extremum** and **Find Anomalies**? Give a concrete example of each from the same dataset.

## References

- Gotz, D., and M. X. Zhou. 2008. "Characterizing Users' Visual Analytic Activity for Insight Provenance." In *Proceedings of IEEE VAST*, 123–130.
- Amar, R., J. Eagan, and J. Stasko. 2005. "Low-Level Components of Analytic Activity in Information Visualization." In *Proceedings of IEEE InfoVis*, 111–117.
- Brehmer, M., and T. Munzner. 2013. "A Multi-Level Typology of Abstract Visualization Tasks." *IEEE Transactions on Visualization and Computer Graphics* 19 (12): 2376–2385.
