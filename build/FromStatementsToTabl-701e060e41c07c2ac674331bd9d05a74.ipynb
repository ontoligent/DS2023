## A more complex example

Let's go through this process again, with a slightly more complex example.

Consider the following descriptive statement, a sentence you might find in a report on global health.

> "In 1970, Germany spent $USD 252.311 per capita and had a life expectany of 70.6 years."

Now, this sentence can be decomposed into a set of atomic facts.

We might write down these facts like this:

```
Observation1 country Germany
Observation1 year 1970
Observation1 spending 252.311
Observation1 life_expectancy 70.6
```

Note that in writing these doown we have surfaced something interesting -- the implicit subject of the sentence is the observation. 

This is a little tricky, because it looks like the the subject is the country. But the year 1970 is not a property of Germany!

Now, imagine we have a bunch of these sets of facts, one set for each country in the world and year between 1970 and 2020.

A segment of these data might look like the following multi-line Python string:

````python
data = """
Observation1 country Germany
Observation1 year 1970
Observation1 spending 252.311
Observation1 life_expectancy 70.6
Observation2 country France 	
Observation2 year 1970 	
Observation2 spending 192.143 	
Observation2 life_expectancy 72.2
Observation3 country GreatBritain
Observation3 year 1970
Observation3 spending 123.993
Observation3 life_expectancy 71.9
Observation4 country Japan
Observation4 year 1970
Observation4 spending 150.437
Observation4 life_expectancy 72.0
Observation5 country USA
Observation5 year 1970
Observation5 spending 326.961
Observation5 life_expectancy 70.9 
"""
````

## Putting the facts into a table

Let's use Pandas to convert these facts into a table

We'll use a list comprehension to parse the string into a list of lists and pass that to a data frame constructor.

````python
lol = [row.split() for row in data[1:-1].split("\n")]
lol
````

````python
import pandas as pd

df = pd.DataFrame(lol, columns=['Subject','Predicate','Object'])
df
````

## Reshaping the table

Now, the shape of this table is not how we normally want our data.

It shows how the data set is composed of facts, but it's not useful in this form for data _analysis_.

For one thing, the Object column contains data of different types, which prevents storing them and processing them properly.

One thing we can do is pivot the table into this form:

````python
df1 = df.pivot(index=['Subject'], columns=['Predicate'], values=['Object'])
df1
````

This puts our data into "tidy" form -- each row is a unique observation, and each column is unique feature, or variable.

Let's look at the complete data set for these observations.

We can find them in Seaborn's collection of data sets.

## Getting the full data set

````python
import seaborn as sns

healthexp = sns.load_dataset("healthexp")
healthexp
````

Again, this is a **tidy** table.

- Each row stands for a unique observation.
- Each column stands for a feature or variable.
- Each cell possesses the value of that variable for a given observation.

We'll dive deeper into what a tidy table is later.

## Processing tidy data

When data are in this form, they are easy to process.

Any feature can be inspected for its statistical properties.

````python
print("Minimum Life Expectancy:", healthexp['Life_Expectancy'].min())
print("Maximum Life Expectancy:", healthexp['Life_Expectancy'].max())
print("Median Life Expectancy:", healthexp['Life_Expectancy'].median())
````

It's also easy to plot. Here are histograms of the two key features:

````python
healthexp['Life_Expectancy'].plot.hist();
````

````python
healthexp['Spending_USD'].plot.hist();
````

And here they are compared in a scatter plot:

````python
healthexp.plot.scatter(x='Spending_USD', y='Life_Expectancy', c='Country');
````

## Creating wide tables

One useful way to prepare a table for analysis is to create "wide" tables -- tables that take a single feature and project its values onto an axis.

Here, we use Pandas' `.pivot()` method to take the `Spending_USD` feature and create a conveniently shaped table that shows how this feature varies with `Country` and `Year`.

````python
healthexp_spending = healthexp.pivot(index="Year", columns="Country", values="Spending_USD")
healthexp_spending.head(10).fillna('')
````

Notice how now, unlike a tidy table, all the cells have the same kind of data -- the come from the same domain.

Tables in this form are convenient for analysis and visualization.

For example, it's easy to get aggregate values and plot them:

````python
healthexp_spending.sum()
````

````python
healthexp_spending.sum().sort_values().plot.barh();
````

````python
healthexp_spending.sum(axis=1).head()
````

````python
healthexp_spending.sum(axis=1).plot.line();
````

````python
healthexp_spending.plot();
````

We can do this with the other feature, too:

````python
healthexp_lifexp = healthexp.pivot(index="Year", columns="Country", values="Life_Expectancy")
healthexp_lifexp.head(10).fillna('')
````

````python
healthexp_lifexp.mean().sort_values().plot.bar();
````

````python
healthexp_lifexp.plot();
````

## Review

In this notebook, we demonstrated the following:

- Descriptive statements can be converted into **atomic facts**.
  - An atomic facts is a simple sentence with a **subject, predicate, and object**.
  - Sometimes the subject is implicit.
- Sets of atomic facts can be represented as **tables**.
  - Tables help us process the data.
- Tables have three basic shapes:
  - **Tables of facts**, where each row is triple of subject, predicate, and object.
  - **Tidy tables** of observations as rows, features as columns, and values as cells.
  - **Wide tables**, created by pivoting a featue from a tidy table.
