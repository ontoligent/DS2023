## Project Overview

This is a **group project** with multiple deliverables spread across the second half of the semester.

You will:

1. Build an **interactive dashboard** exploring a dataset of your own choosing
2. Distill your main finding into a standalone **infographic**

## Dataset Requirements

You may use any dataset, as long as it meets **all** of the following criteria:

| Requirement | Detail |
|---|---|
| Format | Tidy, or convertible to tidy form |
| Columns | At least 3 grouping + 3 measurement columns |
| Size | At least 500 observations |
| Source | Not a built-in dataset (e.g., Seaborn) |

> If you have questions about datasets, ask the instructor.

## Deliverables

Four deliverables — each with its own Canvas submission page and rubric.

1. Project Proposal
2. Dashboard Mockups
3. Final Presentations
4. Final Submission

> Review each deliverable's Canvas page for full instructions.

## Deliverable 1: Project Proposal

A short written proposal describing:

- Your **dataset**
- Your **motivating research questions**
- Your **planned approach**

## Deliverable 2: Dashboard Mockups

- **Low-fidelity mockups** of your planned dashboard
- Workshopped **in class** for peer and instructor feedback
- Feedback informs your implementation before the final presentation

## Deliverable 3: Final Presentations

- **10-minute slot** during the last full week of class (after Thanksgiving)
- Two things presented together:
  1. Implemented dashboard
  2. Infographic draft

Slides from your presentation must be included in your final submission.

### Dashboard: Required Sections

**About Section**
- Overview of the dataset and project goals
- A few static visualizations highlighting key insights
- Acts like a brief introductory article — catches attention, sets context

**Interactive Section**
- At least **three charts** that meaningfully answer your research questions
- Each chart must include some interactivity (widgets, dropdowns, filters, hover, zoom)
- Think carefully about whether interactivity adds real value for each chart

## Deliverable 4: Final Submission

A **standalone infographic** communicating your single main takeaway from the dashboard analysis.

- Combines text and chart(s) to tell the story clearly
- Designed for someone who won't interact with the data themselves
- One chart or several — whatever best supports the takeaway

This is the **revised** version of the infographic from your presentation, updated based on feedback received.
